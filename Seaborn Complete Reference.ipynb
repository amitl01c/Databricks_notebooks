```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Set Seaborn theme
sns.set_theme(style="whitegrid")
```

## 2. Base Plots (Distributions & Relationships)

### Distribution Plots:
* **`sns.histplot(data, x)`** - Histogram
* **`sns.kdeplot(data, x)`** - Kernel density estimate
* **`sns.displot(data, x, kind)`** - Distribution plot (kind='hist', 'kde', 'ecdf')
* **`sns.rugplot(data, x)`** - Rug plot on axis
* **`sns.boxplot(data, x)`** - Boxplot (single variable)
* **`sns.violinplot(data, x)`** - Violin plot
* **`sns.histplot(data, x, hue)`** - Histogram split by category

### Relationship Plots:
* **`sns.scatterplot(data, x, y)`** - Scatter plot
* **`sns.lineplot(data, x, y)`** - Line plot
* **`sns.relplot(data, x, y, kind='scatter')`** - Relationship plot with facets
* **`sns.lmplot(data, x, y)`** - Scatter with regression line

## 3. Categorical Plots

### Plot Types:
* **`sns.barplot(data, x, y)`** - Aggregate bar plot
* **`sns.countplot(data, x)`** - Counts of categorical data
* **`sns.boxplot(data, x, y)`** - Box plot
* **`sns.violinplot(data, x, y)`** - Violin plot
* **`sns.stripplot(data, x, y)`** - Scatter on categorical axis
* **`sns.swarmplot(data, x, y)`** - Non-overlapping scatter on categorical axis
* **`sns.catplot(data, x, y, kind)`** - High-level categorical plot (kind='bar', 'box', etc.)

## 4. Matrix & Pair Plots

### Matrix Plots:
* **`sns.heatmap(data)`** - Heatmap of matrix values
* **`sns.clustermap(data)`** - Hierarchical clustering heatmap

### Pair/Multivariate Plots:
* **`sns.pairplot(data)`** - Pairwise relationships
* **`sns.jointplot(data, x, y, kind)`** - Joint distributions and relationships
* **`sns.corrplot(data)`** - Correlation matrix heatmap

### Example:
* **`sns.pairplot(df, hue='species')`** - Pairwise across groups
