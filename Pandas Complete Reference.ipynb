## 1. Data Structures

### Series (1-dimensional):
* **`pd.Series(data, index)`** - Create a Series from list, dict, or array
* **`s.values`** - Get values as NumPy array
* **`s.index`** - Get index labels
* **`s.dtype`** - Data type of Series
* **`s.name`** - Name of the Series

### DataFrame (2-dimensional):
* **`pd.DataFrame(data, columns, index)`** - Create DataFrame from dict, list, or array
* **`df.shape`** - Dimensions (rows, columns)
* **`df.columns`** - Column names
* **`df.index`** - Row index
* **`df.dtypes`** - Data types of each column
* **`df.info()`** - Summary of DataFrame structure
* **`df.describe()`** - Statistical summary of numeric columns

## 2. Data Loading and Saving

### Reading Data:
* **`pd.read_csv(filepath)`** - Read CSV file
* **`pd.read_excel(filepath)`** - Read Excel file
* **`pd.read_json(filepath)`** - Read JSON file
* **`pd.read_sql(query, connection)`** - Read from SQL database
* **`pd.read_parquet(filepath)`** - Read Parquet file
* **`pd.read_html(url)`** - Read HTML tables from webpage

### Writing Data:
* **`df.to_csv(filepath, index=False)`** - Write to CSV
* **`df.to_excel(filepath, index=False)`** - Write to Excel
* **`df.to_json(filepath)`** - Write to JSON
* **`df.to_sql(table_name, connection)`** - Write to SQL database
* **`df.to_parquet(filepath)`** - Write to Parquet

## 3. Data Inspection and Selection

### Inspection:
* **`df.head(n)`** - First n rows (default 5)
* **`df.tail(n)`** - Last n rows (default 5)
* **`df.sample(n)`** - Random n rows
* **`df.info()`** - Data types and non-null counts
* **`df.describe()`** - Statistical summary
* **`df.shape`** - Number of rows and columns
* **`df.columns`** - Column names
* **`df.index`** - Index values

### Selection:
* **`df['col']`** or **`df.col`** - Select single column (returns Series)
* **`df[['col1', 'col2']]`** - Select multiple columns (returns DataFrame)
* **`df.loc[row_label, col_label]`** - Select by label
* **`df.iloc[row_index, col_index]`** - Select by integer position
* **`df[df['col'] > value]`** - Boolean filtering
* **`df.query('col > value')`** - SQL-like filtering

## 4. Data Cleaning

### Missing Data:
* **`df.isnull()`** or **`df.isna()`** - Detect missing values
* **`df.notnull()`** - Detect non-missing values
* **`df.dropna()`** - Remove rows with missing values
* **`df.fillna(value)`** - Fill missing values
* **`df.ffill()`** - Forward fill missing values
* **`df.bfill()`** - Backward fill missing values
* **`df.interpolate()`** - Interpolate missing values

### Duplicates:
* **`df.duplicated()`** - Identify duplicate rows
* **`df.drop_duplicates()`** - Remove duplicate rows

### Data Type Conversion:
* **`df.astype(dtype)`** - Convert data type
* **`pd.to_numeric()`** - Convert to numeric
* **`pd.to_datetime()`** - Convert to datetime
* **`df['col'].str.strip()`** - Remove whitespace from strings

## 5. Data Manipulation

### Adding/Removing Columns:
* **`df['new_col'] = values`** - Add new column
* **`df.drop('col', axis=1)`** - Remove column
* **`df.rename(columns={'old': 'new'})`** - Rename columns

### Sorting:
* **`df.sort_values('col')`** - Sort by column values
* **`df.sort_index()`** - Sort by index
* **`df.sort_values(['col1', 'col2'], ascending=[True, False])`** - Multi-column sort

### Applying Functions:
* **`df.apply(func)`** - Apply function to columns/rows
* **`df['col'].map(func)`** - Apply function to Series
* **`df.applymap(func)`** - Apply function to every element

### String Operations:
* **`df['col'].str.lower()`** - Convert to lowercase
* **`df['col'].str.contains('pattern')`** - Check if contains pattern
* **`df['col'].str.replace('old', 'new')`** - Replace substring
* **`df['col'].str.split('delimiter')`** - Split strings

## 6. Aggregation and GroupBy

### Aggregation Functions:
* **`df.sum()`** - Sum of values
* **`df.mean()`** - Mean of values
* **`df.median()`** - Median of values
* **`df.min()`** / **`df.max()`** - Minimum/Maximum values
* **`df.std()`** / **`df.var()`** - Standard deviation/Variance
* **`df.count()`** - Count non-null values
* **`df.nunique()`** - Count unique values

### GroupBy Operations:
* **`df.groupby('col').sum()`** - Group and aggregate
* **`df.groupby(['col1', 'col2']).mean()`** - Multi-column grouping
* **`df.groupby('col').agg({'col1': 'sum', 'col2': 'mean'})`** - Multiple aggregations
* **`df.groupby('col').size()`** - Count rows per group
* **`df.pivot_table(values, index, columns, aggfunc)`** - Create pivot table

## 7. Merging and Joining

### Combining DataFrames:
* **`pd.concat([df1, df2])`** - Concatenate DataFrames vertically
* **`pd.concat([df1, df2], axis=1)`** - Concatenate horizontally
* **`df1.append(df2)`** - Append rows (deprecated, use concat)

### Merging (SQL-style joins):
* **`pd.merge(df1, df2, on='key')`** - Inner join
* **`pd.merge(df1, df2, on='key', how='left')`** - Left join
* **`pd.merge(df1, df2, on='key', how='right')`** - Right join
* **`pd.merge(df1, df2, on='key', how='outer')`** - Outer join
* **`df1.join(df2)`** - Join on index

## 8. Time Series Operations

### DateTime Operations:
* **`pd.to_datetime(df['col'])`** - Convert to datetime
* **`df['date'].dt.year`** - Extract year
* **`df['date'].dt.month`** - Extract month
* **`df['date'].dt.day`** - Extract day
* **`df['date'].dt.dayofweek`** - Day of week (0=Monday)
* **`df['date'].dt.weekday`** - Day of week

### Time-based Indexing:
* **`df.set_index('date')`** - Set datetime column as index
* **`df['2024']`** - Select by year
* **`df['2024-01':'2024-03']`** - Select date range
* **`df.resample('D').mean()`** - Resample to daily frequency
* **`df.rolling(window=7).mean()`** - Rolling window calculations
* **`df.shift(1)`** - Shift data by n periods

### Scenario 1: Customer Purchase Analysis
**Problem:** Analyze customer purchase data to identify top customers, revenue trends, and purchase patterns.

```python
import pandas as pd
import numpy as np
```

```python
# Scenario 1: Customer Purchase Analysis
data = {
    'customer_id': [101, 102, 103, 101, 104, 102, 105, 103, 101, 104],
    'product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Laptop', 'Keyboard', 'Mouse', 'Monitor', 'Keyboard', 'Mouse'],
    'amount': [1200, 25, 75, 300, 1200, 75, 25, 300, 75, 25],
    'date': pd.date_range('2026-06-01', periods=10, freq='D')
}
df = pd.DataFrame(data)

print("Purchase Data:")
print(df)

# Total revenue per customer
revenue_by_customer = df.groupby('customer_id')['amount'].sum().sort_values(ascending=False)
print("\nRevenue by Customer:")
print(revenue_by_customer)

# Most popular products
product_counts = df['product'].value_counts()
print("\nProduct Popularity:")
print(product_counts)

# Average transaction value
avg_transaction = df['amount'].mean()
print(f"\nAverage Transaction Value: ${avg_transaction:.2f}")

# Top customer
top_customer = revenue_by_customer.index[0]
top_revenue = revenue_by_customer.iloc[0]
print(f"\nTop Customer: {top_customer} with ${top_revenue:.2f} in purchases")
```

### Scenario 2: Employee Data Cleaning
**Problem:** Clean messy employee data by handling missing values, duplicates, and data type issues.

```python
# Scenario 2: Employee Data Cleaning
employees = pd.DataFrame({
    'name': ['John Doe', 'Jane Smith', 'Bob Jones', 'John Doe', 'Alice Brown', None],
    'age': ['30', '25', None, '30', '35', '28'],
    'salary': [75000, 65000, 80000, 75000, 90000, 70000],
    'department': ['Sales', 'Marketing', 'IT', 'Sales', 'IT', 'Marketing'],
    'email': ['john@company.com', 'jane@company.com', None, 'john@company.com', 'alice@company.com', 'bob@company.com']
})

print("Original Data:")
print(employees)
print(f"\nShape: {employees.shape}")
print(f"Missing values:\n{employees.isnull().sum()}")

# Step 1: Remove duplicates

employees_clean = employees.drop_duplicates()
print(f"\nAfter removing duplicates: {employees_clean.shape}")

# Step 2: Handle missing values
employees_clean = employees_clean.dropna(subset=['name'])  # Remove rows with no name
employees_clean['age'] = employees_clean['age'].fillna(employees_clean['age'].median())  # Fill missing age
employees_clean['email'] = employees_clean['email'].fillna('unknown@company.com')  # Fill missing email

# Step 3: Convert data types
employees_clean['age'] = pd.to_numeric(employees_clean['age'])

print("\nCleaned Data:")
print(employees_clean)
print(f"\nNo missing values: {employees_clean.isnull().sum().sum() == 0}")
```

### Scenario 3: Sales Performance by Region
**Problem:** Aggregate sales data by region and product category to identify performance patterns.

```python
# Scenario 3: Sales Performance by Region
sales_data = pd.DataFrame({
    'region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West'] * 3,
    'category': ['Electronics', 'Electronics', 'Electronics', 'Electronics',
                 'Clothing', 'Clothing', 'Clothing', 'Clothing'] * 3,
    'revenue': [50000, 45000, 60000, 55000, 30000, 35000, 28000, 32000,
                52000, 48000, 62000, 58000, 31000, 36000, 29000, 33000,
                51000, 46000, 61000, 56000, 30500, 35500, 28500, 32500],
    'quarter': ['Q1']*8 + ['Q2']*8 + ['Q3']*8
})

print("Sales Data:")
print(sales_data.head(10))

# Total revenue by region
revenue_by_region = sales_data.groupby('region')['revenue'].sum().sort_values(ascending=False)
print("\nTotal Revenue by Region:")
print(revenue_by_region)

# Average revenue by category
avg_by_category = sales_data.groupby('category')['revenue'].mean()
print("\nAverage Revenue by Category:")
print(avg_by_category)

# Pivot table: Region vs Category
pivot = sales_data.pivot_table(values='revenue', index='region', columns='category', aggfunc='sum')
print("\nRevenue Pivot: Region vs Category")
print(pivot)

# Best performing region-category combination
best_combo = sales_data.groupby(['region', 'category'])['revenue'].sum().idxmax()
print(f"\nBest performing combination: {best_combo[0]} - {best_combo[1]}")
```

### Scenario 4: Website Traffic Analysis
**Problem:** Analyze website traffic over time, calculate rolling averages, and identify trends.

```python
# Scenario 4: Website Traffic Analysis
np.random.seed(42)
dates = pd.date_range('2026-01-01', periods=60, freq='D')
traffic = pd.DataFrame({
    'date': dates,
    'visitors': np.random.randint(1000, 5000, 60) + np.linspace(0, 1000, 60)
})
traffic['visitors'] = traffic['visitors'].astype(int)

print("Traffic Data (first 10 days):")
print(traffic.head(10))

# Set date as index
traffic.set_index('date', inplace=True)

# Calculate 7-day rolling average
traffic['rolling_avg_7d'] = traffic['visitors'].rolling(window=7).mean()

# Calculate daily change
traffic['daily_change'] = traffic['visitors'].diff()

print("\nWith Rolling Average and Daily Change:")
print(traffic.head(10))

# Monthly statistics
monthly_stats = traffic['visitors'].resample('ME').agg(['mean', 'sum', 'max', 'min'])
print("\nMonthly Statistics:")
print(monthly_stats)

# Growth rate (first vs last week)
first_week = traffic['visitors'].iloc[:7].mean()
last_week = traffic['visitors'].iloc[-7:].mean()
growth_rate = ((last_week - first_week) / first_week * 100)
print(f"\nTraffic growth rate: {growth_rate:.1f}%")
```
