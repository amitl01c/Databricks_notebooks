## 1. Basic Plot Types

### Line Plots:
* **`plt.plot(x, y)`** - Basic line plot
* **`plt.plot(x, y, 'r--')`** - Red dashed line
* **`plt.plot(x, y, marker='o')`** - Line with markers

### Scatter Plots:
* **`plt.scatter(x, y)`** - Scatter plot
* **`plt.scatter(x, y, s=sizes, c=colors)`** - With size and color

### Bar Charts:
* **`plt.bar(x, height)`** - Vertical bar chart
* **`plt.barh(y, width)`** - Horizontal bar chart

### Histograms:
* **`plt.hist(data, bins=20)`** - Histogram
* **`plt.hist(data, bins=20, density=True)`** - Normalized histogram

### Pie Charts:
* **`plt.pie(sizes, labels=labels)`** - Pie chart
* **`plt.pie(sizes, labels=labels, autopct='%1.1f%%')`** - With percentages

## 2. Figure and Axes

### Creating Figures:
* **`fig, ax = plt.subplots()`** - Create figure and single axis
* **`fig, axes = plt.subplots(nrows, ncols)`** - Create multiple subplots
* **`fig = plt.figure(figsize=(10, 6))`** - Create figure with custom size

### Axes Methods:
* **`ax.plot(x, y)`** - Plot on specific axis
* **`ax.set_xlabel('label')`** - Set x-axis label
* **`ax.set_ylabel('label')`** - Set y-axis label
* **`ax.set_title('title')`** - Set plot title
* **`ax.set_xlim(xmin, xmax)`** - Set x-axis limits
* **`ax.set_ylim(ymin, ymax)`** - Set y-axis limits
* **`ax.grid(True)`** - Add grid
* **`ax.legend()`** - Add legend

## 3. Customization

### Colors:
* Single letter: `'r'` (red), `'b'` (blue), `'g'` (green), `'k'` (black)
* Named colors: `'red'`, `'blue'`, `'green'`
* Hex colors: `'#FF5733'`
* RGB tuples: `(0.5, 0.5, 0.5)`

### Line Styles:
* **`'-'`** - Solid line
* **`'--'`** - Dashed line
* **`'-.'`** - Dash-dot line
* **`':'`** - Dotted line

### Markers:
* **`'o'`** - Circle
* **`'s'`** - Square
* **`'^'`** - Triangle
* **`'*'`** - Star
* **`'+'`** - Plus

### Legends:
* **`plt.legend()`** - Add legend
* **`plt.legend(loc='upper right')`** - Specify location
* **`plt.legend(labels=['A', 'B'])`** - Custom labels

## 4. Multiple Subplots

### Creating Subplots:
* **`fig, axes = plt.subplots(2, 2)`** - 2x2 grid of subplots
* **`fig, axes = plt.subplots(1, 3, figsize=(15, 5))`** - 1x3 with custom size
* **`plt.subplot(2, 2, 1)`** - Add subplot at position 1

### Accessing Subplots:
* **`axes[0, 0].plot(x, y)`** - Plot on specific subplot (2D array)
* **`axes[i].plot(x, y)`** - Plot on specific subplot (1D array)

### Layout:
* **`plt.tight_layout()`** - Adjust spacing automatically
* **`plt.subplots_adjust(hspace, wspace)`** - Manual spacing adjustment

## 5. Advanced Plot Types

### Box Plots:
* **`plt.boxplot(data)`** - Box plot for distribution

### Violin Plots:
* **`plt.violinplot(data)`** - Violin plot

### Heatmaps:
* **`plt.imshow(matrix, cmap='hot')`** - Display matrix as heatmap
* **`plt.colorbar()`** - Add colorbar

### Contour Plots:
* **`plt.contour(X, Y, Z)`** - Contour lines
* **`plt.contourf(X, Y, Z)`** - Filled contours

### Error Bars:
* **`plt.errorbar(x, y, yerr=errors)`** - Plot with error bars

### Fill Between:
* **`plt.fill_between(x, y1, y2)`** - Fill area between curves

## 6. Annotations and Text

### Adding Text:
* **`plt.text(x, y, 'text')`** - Add text at position
* **`plt.annotate('text', xy=(x, y))`** - Add annotation
* **`plt.annotate('text', xy=(x, y), xytext=(x2, y2), arrowprops=dict())`** - Annotation with arrow

### Formatting:
* **`fontsize=12`** - Font size
* **`fontweight='bold'`** - Font weight
* **`color='red'`** - Text color
* **`ha='center'`** - Horizontal alignment
* **`va='center'`** - Vertical alignment

## 7. Saving Figures

### Save Commands:
* **`plt.savefig('plot.png')`** - Save as PNG
* **`plt.savefig('plot.pdf')`** - Save as PDF
* **`plt.savefig('plot.svg')`** - Save as SVG
* **`plt.savefig('plot.png', dpi=300)`** - Save with high resolution
* **`plt.savefig('plot.png', bbox_inches='tight')`** - Save without extra whitespace
* **`plt.savefig('plot.png', transparent=True)`** - Save with transparent background

###1) Line plot
A line chart is used to show the change in information over time. The horizontal axis is usually a time scale; for example, minutes, hours,
days, months, or years.
plt. Use the .plot() method and provide data to create a plot. Then, use the .show() method to display the plot

```python
plt.plot(year,sales) 
plt.show() 
```

### Add chart elements
-  plt.title(): Sets the title of the chart, which is passed as an argument.
- plt.xlabel(): Sets the label of the X axis.
- plt.ylabel(): Sets the label of the Y axis.
- plt.grid(): for adding gridlines.
- plt.show(): for displaying the plot

```python
plt.plot(year,sales) 
plt.title("Year Wise Sales Trend") 
plt.xlabel("Year") 
plt.ylabel("Sales") 
plt.grid(axis="y") 
plt.show()
```

```python
plt.plot(year,sales,color="red",linestyle="--",marker="o",markersize=5) 
plt.title("Year Wise Sales Trend") 
plt.xlabel("Year") 
plt.ylabel("Sales") 
plt.grid(axis="y") 
plt.show()
```

### 3. Bar chart
A bar chart is used when you want to show a distribution of data points or perform a comparison of metric values across different
subgroups of your data. From a bar chart, we can see which groups are highest or most common, and how other groups compare against
the others.c values.
In a bar chart, we have one axis representing a particular category of the columns and another axis representing the values or count of
the specific category.
Examples:- Total sales by product category 
 - population by country 
 - Revenue by department

```python
players = ["Dhoni","Kohli", "Sachin", "Jadeja", "KL"] 
runs = [2000,3000, 4000, 2500, 2000] 
```

```python
plt.bar(x=players,height=runs) 
plt.show() 
```

```python
name = ["sachin verma","shubham patel","rahul patidar","Devendra patidar","Lalit mandloi"] 
sales = [15000, 12000, 20000, 13000,14000] 
plt.bar(name,sales) 
plt.show()  
```

```python
name = ["sachin verma","shubham patel","rahul patidar","Devendra patidar","Lalit mandloi"] 
sales = [15000, 12000, 20000, 13000,14000] 
plt.bar(name,sales) 
plt.xticks(rotation="vertical")  # use rotation parameter to set the labels vertical or horizontal 
plt.show()  
```

**horizontal bar**
A horizontal bar chart is a great option for long category names, because there is more space on the left-hand side of the chart for axis
labels to be placed and horizontally oriented.
A horizontal bar chart would be a better choice if the text on the x-axis of a vertical bar chart would have to be diagonal (or worse, cut
off) to fit.
plt.barh()

```python
plt.barh(players,runs,color="red") 
plt.show()
```

### 5) Pie Chart
A pie chart is a type of graph in which a circle is divided into sectors that each represents a proportion of the whole(100%).
Examples:
Percentage of budget spent by department
Gender distribution
Favorite Type of Movie

```python
genre = ["Comedy","Action","Romance","Drama","SciFi"] 
no_of_movies = [1000,1500,1200,900,800] 
plt.pie(no_of_movies,labels=genre) 
plt.show()
```

```python
# Use this command (autopct='%0.1f%%') to display data in percentages  
plt.pie(no_of_movies,labels=genre,autopct='%0.1f%%') 
plt.show() 
```
