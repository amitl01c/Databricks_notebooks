## 1. Array Creation Functions

### Basic Array Creation:
* **`np.array()`** - Create array from list or tuple
* **`np.arange(start, stop, step)`** - Create array with evenly spaced values
* **`np.linspace(start, stop, num)`** - Create array with specified number of evenly spaced values
* **`np.zeros(shape)`** - Create array filled with zeros
* **`np.ones(shape)`** - Create array filled with ones
* **`np.empty(shape)`** - Create uninitialized array
* **`np.eye(n)`** - Create identity matrix
* **`np.full(shape, fill_value)`** - Create array filled with specified value

## 2. Random Number Generation

### Random Functions:
* **`np.random.rand(d0, d1, ...)`** - Random values in [0, 1) with uniform distribution
* **`np.random.randn(d0, d1, ...)`** - Random values from standard normal distribution
* **`np.random.randint(low, high, size)`** - Random integers from low (inclusive) to high (exclusive)
* **`np.random.random(size)`** - Random floats in [0.0, 1.0)
* **`np.random.choice(a, size)`** - Random sample from given array
* **`np.random.shuffle(a)`** - Shuffle array in-place
* **`np.random.seed(seed)`** - Set random seed for reproducibility

## 3. Array Attributes

### Important Array Properties:
* **`arr.dtype`** - Data type of array elements (int32, float64, etc.)
* **`arr.ndim`** - Number of dimensions (1D, 2D, 3D, etc.)
* **`arr.shape`** - Tuple of array dimensions (rows, columns)
* **`arr.size`** - Total number of elements
* **`arr.itemsize`** - Size in bytes of each element
* **`arr.nbytes`** - Total bytes consumed by array (size × itemsize)
* **`arr.T`** - Transpose of the array
* **`arr.real`** - Real part of complex array
* **`arr.imag`** - Imaginary part of complex array

## 4. Array Indexing and Slicing

### Indexing Methods:
* **`arr[i]`** - Access element at index i (1D) or row i (2D)
* **`arr[i, j]`** - Access element at row i, column j
* **`arr[i][j]`** - Alternative 2D indexing (less efficient)
* **`arr[start:stop:step]`** - Slice array with start, stop, step
* **`arr[:, j]`** - Access entire column j
* **`arr[i, :]`** - Access entire row i
* **`arr[arr > 5]`** - Boolean indexing (filter elements)
* **`arr[[1, 3, 5]]`** - Fancy indexing (select specific indices)

### Scenario 1: Student Grade Analysis
**Problem:** Analyze student exam scores to calculate statistics and identify top performers.

```python
# Scenario 1: Student Grade Analysis
# Student scores in 3 exams (rows = students, columns = exams)
student_scores = np.array([
    [85, 92, 78],  # Student 1
    [90, 88, 95],  # Student 2
    [76, 84, 80],  # Student 3
    [95, 91, 89],  # Student 4
    [82, 79, 85]   # Student 5
])

print("Student Scores:\n", student_scores)
print("\nAverage score per student:", np.mean(student_scores, axis=1).round(2))
print("Average score per exam:", np.mean(student_scores, axis=0).round(2))
print("\nHighest score:", np.max(student_scores))
print("Lowest score:", np.min(student_scores))
print("Overall average:", np.mean(student_scores).round(2))

# Find top performer
top_student_idx = np.argmax(np.mean(student_scores, axis=1))
print(f"\nTop performer: Student {top_student_idx + 1} with average {np.mean(student_scores[top_student_idx]).round(2)}")

# Students scoring above 85
high_achievers = student_scores >= 85
print(f"\nNumber of scores >= 85: {np.sum(high_achievers)}")
```

### Scenario 2: Sales Data Analysis
**Problem:** Analyze monthly sales data to identify trends, calculate growth, and find best/worst performing months.

```python
# Scenario 2: Sales Data Analysis
# Monthly sales for 4 products over 6 months
monthly_sales = np.array([
    [12000, 15000, 13500, 16000, 18000, 20000],  # Product A
    [8000, 8500, 9000, 9200, 9800, 10500],       # Product B
    [5000, 5200, 4800, 6000, 6500, 7000],        # Product C
    [15000, 14500, 16000, 17000, 18500, 19000]   # Product D
])

products = ['Product A', 'Product B', 'Product C', 'Product D']
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']

print("Monthly Sales Data:\n", monthly_sales)
print("\n--- Analysis ---")

# Total sales per product
total_per_product = np.sum(monthly_sales, axis=1)
for i, product in enumerate(products):
    print(f"{product} total sales: ${total_per_product[i]:,}")

# Best and worst months
total_per_month = np.sum(monthly_sales, axis=0)
best_month_idx = np.argmax(total_per_month)
worst_month_idx = np.argmin(total_per_month)
print(f"\nBest performing month: {months[best_month_idx]} (${total_per_month[best_month_idx]:,})")
print(f"Worst performing month: {months[worst_month_idx]} (${total_per_month[worst_month_idx]:,})")

# Growth rate (comparing first and last month)
growth_rate = ((monthly_sales[:, -1] - monthly_sales[:, 0]) / monthly_sales[:, 0] * 100)
print("\nGrowth rate (Jan to Jun):")
for i, product in enumerate(products):
    print(f"{product}: {growth_rate[i]:.1f}%")
```

### Scenario 3: Temperature Data Analysis
**Problem:** Analyze daily temperature readings, identify outliers, and calculate statistics for weather patterns.

```python
# Scenario 3: Temperature Data Analysis
# Daily temperatures (°F) for a week
daily_temps = np.array([72, 75, 78, 105, 74, 76, 73])  # Day 4 has an outlier

print("Daily Temperatures:", daily_temps)
print("\n--- Statistics ---")
print(f"Mean temperature: {np.mean(daily_temps):.1f}°F")
print(f"Median temperature: {np.median(daily_temps):.1f}°F")
print(f"Std deviation: {np.std(daily_temps):.1f}°F")

# Detect outliers (values > 2 standard deviations from mean)
mean_temp = np.mean(daily_temps)
std_temp = np.std(daily_temps)
threshold = 2 * std_temp
outliers = np.abs(daily_temps - mean_temp) > threshold

print(f"\nOutliers detected: {daily_temps[outliers]}")
print(f"Outlier days: Day {np.where(outliers)[0] + 1}")

# Clean data by removing outliers
clean_temps = daily_temps[~outliers]
print(f"\nCleaned data: {clean_temps}")
print(f"Mean after cleaning: {np.mean(clean_temps):.1f}°F")
print(f"Median after cleaning: {np.median(clean_temps):.1f}°F")
```

### Scenario 4: Portfolio Investment Returns
**Problem:** Calculate investment returns, risk metrics, and portfolio performance over time.

```python
# Scenario 4: Portfolio Investment Returns
# Monthly returns (%) for 3 stocks over 12 months
stock_returns = np.array([
    [2.5, -1.2, 3.1, 1.8, -0.5, 2.2, 1.5, 3.0, -1.0, 2.8, 1.2, 2.5],  # Stock A
    [1.8, 2.1, -2.0, 2.5, 1.3, -0.8, 2.0, 1.5, 2.2, -1.5, 3.0, 1.8],  # Stock B
    [3.2, 1.5, 2.8, -1.5, 2.0, 2.5, -0.5, 2.8, 1.8, 2.0, 1.5, 2.2]   # Stock C
])

stock_names = ['Stock A', 'Stock B', 'Stock C']

print("Monthly Returns (%) for Each Stock:\n", stock_returns)
print("\n--- Performance Metrics ---")

# Average return and volatility (risk) for each stock
for i, name in enumerate(stock_names):
    avg_return = np.mean(stock_returns[i])
    volatility = np.std(stock_returns[i])
    total_return = np.sum(stock_returns[i])
    print(f"\n{name}:")
    print(f"  Average monthly return: {avg_return:.2f}%")
    print(f"  Volatility (risk): {volatility:.2f}%")
    print(f"  Total annual return: {total_return:.2f}%")
    print(f"  Risk-adjusted return: {avg_return/volatility:.3f}")

# Portfolio analysis (equal weight)
portfolio_weights = np.array([1/3, 1/3, 1/3])
portfolio_returns = np.dot(portfolio_weights, stock_returns)
print(f"\nPortfolio Performance (Equal Weight):")
print(f"  Average monthly return: {np.mean(portfolio_returns):.2f}%")
print(f"  Portfolio volatility: {np.std(portfolio_returns):.2f}%")
print(f"  Total annual return: {np.sum(portfolio_returns):.2f}%")
```

### Scenario 5: Matrix Operations for Data Transformation
**Problem:** Apply matrix transformations for data normalization and feature scaling in machine learning.

```python
# Scenario 5: Matrix Operations for Data Transformation
# Raw feature data: [Age, Income, Credit Score]
raw_data = np.array([
    [25, 45000, 680],
    [35, 75000, 720],
    [45, 95000, 750],
    [28, 52000, 690],
    [52, 120000, 780]
])

print("Raw Data (Age, Income, Credit Score):\n", raw_data)

# Min-Max Normalization (scale to 0-1)
def min_max_normalize(data):
    min_vals = np.min(data, axis=0)
    max_vals = np.max(data, axis=0)
    return (data - min_vals) / (max_vals - min_vals)

normalized = min_max_normalize(raw_data)
print("\nNormalized Data (0-1 scale):\n", normalized.round(3))

# Z-score Standardization (mean=0, std=1)
def standardize(data):
    mean_vals = np.mean(data, axis=0)
    std_vals = np.std(data, axis=0)
    return (data - mean_vals) / std_vals

standardized = standardize(raw_data)
print("\nStandardized Data (Z-scores):\n", standardized.round(3))

# Verify standardization
print("\nVerification - Mean of standardized data:", np.mean(standardized, axis=0).round(10))
print("Verification - Std of standardized data:", np.std(standardized, axis=0).round(3))
```

### Scenario 6: Image Representation and Manipulation
**Problem:** Represent and manipulate images as NumPy arrays (RGB channels, grayscale conversion, brightness adjustment).

```python
# Scenario 6: Image Representation and Manipulation
# Simulated 3x3 RGB image (height, width, channels)
# Each pixel has 3 values: Red, Green, Blue (0-255)
image_rgb = np.array([
    [[255, 0, 0], [0, 255, 0], [0, 0, 255]],     # Row 1: Red, Green, Blue pixels
    [[255, 255, 0], [255, 0, 255], [0, 255, 255]], # Row 2: Yellow, Magenta, Cyan
    [[128, 128, 128], [255, 255, 255], [0, 0, 0]]  # Row 3: Gray, White, Black
])

print("RGB Image Array Shape:", image_rgb.shape)
print("Image (3x3 pixels, RGB):\n", image_rgb)

# Extract individual color channels
red_channel = image_rgb[:, :, 0]
green_channel = image_rgb[:, :, 1]
blue_channel = image_rgb[:, :, 2]

print("\nRed Channel:\n", red_channel)
print("Green Channel:\n", green_channel)
print("Blue Channel:\n", blue_channel)

# Convert to grayscale using weighted average
# Standard formula: 0.299*R + 0.587*G + 0.114*B
grayscale = np.dot(image_rgb[..., :3], [0.299, 0.587, 0.114])
print("\nGrayscale Image:\n", grayscale.round(1))

# Increase brightness by 50 (capped at 255)
brightened = np.clip(image_rgb + 50, 0, 255)
print("\nBrightened Image (first pixel):", brightened[0, 0])
print("Original (first pixel):", image_rgb[0, 0])
```

### Scenario 7: Probability Simulation - Dice Rolls
**Problem:** Simulate dice rolls to calculate probabilities and verify theoretical expectations.

```python
# Scenario 7: Probability Simulation - Dice Rolls
# Simulate rolling two dice 10,000 times
np.random.seed(42)  # For reproducibility
n_rolls = 10000

# Roll two dice
dice1 = np.random.randint(1, 7, n_rolls)
dice2 = np.random.randint(1, 7, n_rolls)
total = dice1 + dice2

print(f"Simulating {n_rolls:,} rolls of two dice")
print(f"\nFirst 10 rolls:")
print(f"Die 1: {dice1[:10]}")
print(f"Die 2: {dice2[:10]}")
print(f"Sum:   {total[:10]}")

# Calculate probabilities for each sum (2-12)
print("\n--- Probability Distribution ---")
for sum_value in range(2, 13):
    count = np.sum(total == sum_value)
    probability = count / n_rolls
    # Theoretical probability
    theoretical = {2: 1/36, 3: 2/36, 4: 3/36, 5: 4/36, 6: 5/36, 7: 6/36,
                   8: 5/36, 9: 4/36, 10: 3/36, 11: 2/36, 12: 1/36}
    print(f"Sum = {sum_value:2d}: {probability:.4f} (theoretical: {theoretical[sum_value]:.4f})")

# Statistics
print(f"\nStatistics:")
print(f"Mean sum: {np.mean(total):.2f} (theoretical: 7.00)")
print(f"Most common sum: {np.argmax(np.bincount(total)[2:])+2}")
print(f"Standard deviation: {np.std(total):.2f}")

# Probability of rolling doubles
doubles = np.sum(dice1 == dice2)
print(f"\nDoubles rolled: {doubles} times ({doubles/n_rolls*100:.2f}%, theoretical: 16.67%)")
```

### Scenario 8: Time Series Analysis - Moving Averages
**Problem:** Analyze time series data using rolling windows and moving averages for trend analysis.

```python
# Scenario 8: Time Series Analysis - Moving Averages
# Daily stock prices with noise and trend
np.random.seed(42)
days = 30
trend = np.linspace(100, 150, days)  # Upward trend
noise = np.random.normal(0, 5, days)  # Random fluctuations
stock_prices = trend + noise

print("Stock Prices (first 10 days):", stock_prices[:10].round(2))

# Calculate 5-day moving average
window_size = 5
moving_avg = np.convolve(stock_prices, np.ones(window_size)/window_size, mode='valid')

print(f"\n5-Day Moving Average (first 10 values):", moving_avg[:10].round(2))

# Daily returns (percentage change)
daily_returns = (stock_prices[1:] - stock_prices[:-1]) / stock_prices[:-1] * 100

print("\nDaily Returns (%):", daily_returns[:10].round(2))
print(f"\nStatistics:")
print(f"Starting price: ${stock_prices[0]:.2f}")
print(f"Ending price: ${stock_prices[-1]:.2f}")
print(f"Total return: {((stock_prices[-1] - stock_prices[0]) / stock_prices[0] * 100):.2f}%")
print(f"Average daily return: {np.mean(daily_returns):.2f}%")
print(f"Volatility (std of returns): {np.std(daily_returns):.2f}%")
print(f"Best day return: {np.max(daily_returns):.2f}%")
print(f"Worst day return: {np.min(daily_returns):.2f}%")
```
