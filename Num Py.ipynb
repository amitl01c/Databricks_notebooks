## 1. Array Creation Functions

### Basic Array Creation:
* **`np.array()`** - Create array from list or tuple
* **`np.arange(start, stop, step)`** - Create array with evenly spaced values
* **`np.linspace(start, stop, num)`** - Create array with specified number of evenly spaced values
* **`np.zeros(shape)`** - Create array filled with zeros
* **`np.ones(shape)`** - Create array filled with ones
* **`np.empty(shape)`** - Create uninitialized array
* **`np.eye(n)`** - Create identity matrix
* **`np.full(shape, fill_value)`** - Create array filled with specified value

## 2. Random Number Generation

### Random Functions:
* **`np.random.rand(d0, d1, ...)`** - Random values in [0, 1) with uniform distribution
* **`np.random.randn(d0, d1, ...)`** - Random values from standard normal distribution
* **`np.random.randint(low, high, size)`** - Random integers from low (inclusive) to high (exclusive)
* **`np.random.random(size)`** - Random floats in [0.0, 1.0)
* **`np.random.choice(a, size)`** - Random sample from given array
* **`np.random.shuffle(a)`** - Shuffle array in-place
* **`np.random.seed(seed)`** - Set random seed for reproducibility

## 3. Array Attributes

### Important Array Properties:
* **`arr.dtype`** - Data type of array elements (int32, float64, etc.)
* **`arr.ndim`** - Number of dimensions (1D, 2D, 3D, etc.)
* **`arr.shape`** - Tuple of array dimensions (rows, columns)
* **`arr.size`** - Total number of elements
* **`arr.itemsize`** - Size in bytes of each element
* **`arr.nbytes`** - Total bytes consumed by array (size × itemsize)
* **`arr.T`** - Transpose of the array
* **`arr.real`** - Real part of complex array
* **`arr.imag`** - Imaginary part of complex array

## 4. Array Indexing and Slicing

### Indexing Methods:
* **`arr[i]`** - Access element at index i (1D) or row i (2D)
* **`arr[i, j]`** - Access element at row i, column j
* **`arr[i][j]`** - Alternative 2D indexing (less efficient)
* **`arr[start:stop:step]`** - Slice array with start, stop, step
* **`arr[:, j]`** - Access entire column j
* **`arr[i, :]`** - Access entire row i
* **`arr[arr > 5]`** - Boolean indexing (filter elements)
* **`arr[[1, 3, 5]]`** - Fancy indexing (select specific indices)

## 5. Mathematical Operations

### Element-wise Operations:
* **`np.add(a, b)` or `a + b`** - Addition
* **`np.subtract(a, b)` or `a - b`** - Subtraction
* **`np.multiply(a, b)` or `a * b`** - Multiplication
* **`np.divide(a, b)` or `a / b`** - Division
* **`np.power(a, n)` or `a ** n`** - Power
* **`np.sqrt(a)`** - Square root
* **`np.exp(a)`** - Exponential
* **`np.log(a)`** - Natural logarithm
* **`np.sin(a), np.cos(a), np.tan(a)`** - Trigonometric functions
* **`np.dot(a, b)` or `a @ b`** - Matrix multiplication
* **`np.abs(a)`** - Absolute value

## 6. Statistical Functions

### Aggregate Functions:
* **`np.sum(arr, axis=None)`** - Sum of array elements
* **`np.mean(arr, axis=None)`** - Mean of array elements
* **`np.median(arr, axis=None)`** - Median of array elements
* **`np.std(arr, axis=None)`** - Standard deviation
* **`np.var(arr, axis=None)`** - Variance
* **`np.min(arr, axis=None)`** - Minimum value
* **`np.max(arr, axis=None)`** - Maximum value
* **`np.argmin(arr)`** - Index of minimum value
* **`np.argmax(arr)`** - Index of maximum value
* **`np.percentile(arr, q)`** - q-th percentile
* **`np.corrcoef(x, y)`** - Correlation coefficient

**Note:** Use `axis=0` for column-wise operations, `axis=1` for row-wise operations

## 7. Array Manipulation

### Reshaping and Combining:
* **`arr.reshape(shape)`** - Change array shape without changing data
* **`arr.flatten()`** - Convert to 1D array
* **`arr.ravel()`** - Return flattened array (view if possible)
* **`np.concatenate([a, b], axis)`** - Join arrays along axis
* **`np.vstack([a, b])`** - Stack arrays vertically (row-wise)
* **`np.hstack([a, b])`** - Stack arrays horizontally (column-wise)
* **`np.split(arr, sections, axis)`** - Split array into sections
* **`arr.transpose()` or `arr.T`** - Transpose array
* **`np.expand_dims(arr, axis)`** - Add new axis
* **`np.squeeze(arr)`** - Remove single-dimensional entries

### Sorting and Searching:
* **`np.sort(arr, axis)`** - Sort array
* **`np.argsort(arr)`** - Indices that would sort array
* **`np.where(condition)`** - Return indices where condition is true
* **`np.unique(arr)`** - Find unique elements

## 8. Additional Useful Functions

### Comparison and Logic:
* **`np.equal(a, b)` or `a == b`** - Element-wise equality
* **`np.greater(a, b)` or `a > b`** - Element-wise greater than
* **`np.logical_and(a, b)`** - Element-wise AND
* **`np.logical_or(a, b)`** - Element-wise OR
* **`np.logical_not(a)`** - Element-wise NOT
* **`np.all(arr)`** - Test if all elements are True
* **`np.any(arr)`** - Test if any element is True

### Copy and Type Conversion:
* **`arr.copy()`** - Create deep copy of array
* **`arr.astype(dtype)`** - Convert array to different type
* **`np.isnan(arr)`** - Check for NaN values
* **`np.isinf(arr)`** - Check for infinite values

```python
# Mathematical operations
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 3, 4, 5, 6])

print("Addition:", x + y)
print("Multiplication:", x * y)
print("Power:", x ** 2)
print("Square root:", np.sqrt(x))
print("Sum of all elements:", np.sum(x))
```

```python
# Statistical functions
data = np.array([[10, 20, 30], [40, 50, 60]])

print("Data:\n", data)
print("\nMean:", np.mean(data))
print("Mean by column (axis=0):", np.mean(data, axis=0))
print("Mean by row (axis=1):", np.mean(data, axis=1))
print("\nMax:", np.max(data))
print("Min:", np.min(data))
print("Std deviation:", np.std(data))
```

```python
# Array manipulation
original = np.arange(12)
print("Original array:", original)

reshaped = original.reshape(3, 4)
print("\nReshaped (3x4):\n", reshaped)

transposed = reshaped.T
print("\nTransposed:\n", transposed)

flattened = reshaped.flatten()
print("\nFlattened:", flattened)
```

```python
# Boolean indexing
numbers = np.array([1, 5, 10, 15, 20, 25, 30])
print("Original array:", numbers)

# Filter values greater than 15
filtered = numbers[numbers > 15]
print("Values > 15:", filtered)

# Filter even numbers
even_numbers = numbers[numbers % 2 == 0]
print("Even numbers:", even_numbers)

# Replace values using boolean indexing
numbers_copy = numbers.copy()
numbers_copy[numbers_copy < 10] = 0
print("Replace values < 10 with 0:", numbers_copy)
```

```python
# Array stacking and concatenation
a1 = np.array([[1, 2], [3, 4]])
a2 = np.array([[5, 6], [7, 8]])

print("Array 1:\n", a1)
print("\nArray 2:\n", a2)

vstacked = np.vstack([a1, a2])
print("\nVertical stack:\n", vstacked)

hstacked = np.hstack([a1, a2])
print("\nHorizontal stack:\n", hstacked)
```
